```python
import pandas as pd
df1=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['raja','rani','rahul','rhea'],
    'age':[25,30,35,40]
})
df2=pd.DataFrame({
    'id':[3,4,5,6],
    'name':['rahul','rhea','ravi','reena'],
    'age':[35,40,45,50]
})
df3=pd.concat([df1,df2])
print(df3)


df3=pd.merge(df1,df2,on='id',how='inner')
print(df3)

df4=pd.merge(df1,df2,on='id',how='left')
df4.rename(columns={'name_x':'table_name1'},inplace=True)
print(df3)

df5=pd.merge(df1,df2,how='cross')
display(df5)

```
