```python
lst = [1,2,3,'hello', 'c', [2,4]]
print(lst)
# append()
lst.append(3)
print(lst)

# extend()
lst.extend([2,3,4])
print(lst)

#insert()
lst.insert(2, "value")
print(lst)

# pop()
lst.pop()
print(lst)

# remove (remove value)
lst.remove('hello')
print(lst)

# index()
print(lst.index(2))

# count()
print(lst.count(3))

# sort()
lst = [82,37,45]
print(lst)
lst.sort()
print(lst)

# reverse()
lst = [82,37,45]
print(lst)
lst.reverse()
print(lst)

```
